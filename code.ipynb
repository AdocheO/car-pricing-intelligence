```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
df = pd.read_csv("autos.csv", encoding="latin1")
df.head(5)
```

```python
df.info()
```

```python
df.describe()
```

```python
 df['price'].describe() 
```

```python
df['price'].value_counts().head()
```

```python
df["price"].replace('$', '', inplace=True)
df.drop(df[df['price'] == 0].index, inplace=True)
df.head()
```

```python
df['price'].describe()
```

```python
df['price'] =df['price'].str.replace('$', '', regex=False)
df['price'] =df['price'].str.replace(',', '', regex=False)
df['price'] = df['price'].astype(float)
df['price'].head()
```

```python
df.dropna(inplace=True )
df=df[df['price'] > 0]
```

```python
df['price'].describe()

```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,3))
df['price'].hist(bins=10)

plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Price Distribution")

plt.show()
```

```python
df['price'].quantile([0.90, 0.95, 0.99, 0.999])
df = df[df['price'] < 100000]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,3))

df['price'].hist(bins=50)

plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Price Distribution After Removing Outliers")

plt.show()
```

```python
df['brand'].value_counts().head(20)
```

```python
brand_price = df.groupby('brand')['price'].mean().sort_values(ascending=False)
```

```python
brand_price.head(20)
```

```python
brand_summary = df.groupby('brand').agg({
    'price': ['mean', 'count']
})

brand_summary.columns = ['average_price', 'listing_count']

brand_summary.sort_values(by='average_price', ascending=False).head(20)
```

```python
import matplotlib.pyplot as plt

top_brands = brand_summary.sort_values(
    by='average_price',
    ascending=False
).head(10)

plt.figure(figsize=(10,3))

plt.bar(top_brands.index, top_brands['average_price'])

plt.xticks(rotation=45)

plt.xlabel("Brand")
plt.ylabel("Average Price")
plt.title("Top 10 Brands by Average Price")

plt.show()
```
